#  Mountain Rescue Incident
This notebook consisted of some background analysis behind the web application on https://shiminc.github.io/mountain_rescue.html
- The app is aimed to raise readers' awareness about the times and efforts of mountain rescue team
- Therfore, there are interactive visualisation for readers to explore and the main content is written for general readers rather than research paper. 


## Research Questions
1. When do mountain rescue incidents most likely to occur across the years?
- Is it possible to predict the number of future incidents to assist in planning 
2. Is there any association between the month and the incident cause?
3. How much time and efforts are spent on mountain rescue?
 

## Data 
Web-scraping of [Wasdale Mountain Rescue](https://www.wmrt.org.uk/)
Incident reports since 1968, but patchy and inconsistent reporting until past 10 years. 
Therefore only 2015 - 2025 was used in the analysis. 

## Variables of interest
1. Count of incidents, 
2. Unit of analysis - month, year
3. Incident Cause 
4. hours, staff, staff-hours

## Limitation
1. Data: Potential investigation of variables affecting incident numbers and rescue time
- visitors number, social media, historical weather data 
2. Relative small dataset (1218 incidents)
- mainly descriptive analysis and non parametric analysis
- no separation of raw data and processed data, as it is relatively fast to process the raw data every time each analysis runs.
